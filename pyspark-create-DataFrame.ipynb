```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StringType, StructField, IntegerType, StructType,LongType
```

Initialize a spark session


```python
spark = SparkSession.builder.appName('MyApplication').getOrCreate()
spark
```

Define a Data

```python
data = [("Yuga", 29,"Mississauga",7040565688),
        ("Rushi",15,"Toronto",743998855),
        ("Aarav", 34, "Brampton", 6471234567),  
("Vihaan", 22, "Scarborough", 4169876543),  
("Mira", 27, "Vancouver", 7782345678),  
("Kiran", 31, "Calgary", 4038765432),  
("Tanvi", 19, "Ottawa", 6137654321)  
]
```

Define Schema

```python
schema = StructType([
    StructField("Name", StringType(), True),
    StructField("Age", IntegerType(), True),
    StructField("City", StringType(), True),
    StructField("Mobile", LongType(), True)])
```

```python
df = spark.createDataFrame(data, schema)
df.show()
```
