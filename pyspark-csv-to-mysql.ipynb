```python
from pyspark.sql import SparkSession

# Initialize Spark session with the MySQL JDBC driver from Maven
spark = SparkSession.builder \
    .appName("CSV to MySQL") \
    .config("spark.jars.packages", "mysql:mysql-connector-java:8.0.x") \
    .getOrCreate()
```

```python
csv_file_path = "/workspaces/PySpark/employees.csv"
df = spark.read.csv(csv_file_path, header =True, inferSchema = True)
df. show()
```

```python
# MySQL database connection properties
mysql_url = "jdbc:mysql://<mysql_host>:<mysql_port>/<database_name>"
mysql_properties = {
    "user": "yuga",
    "password": "admin123",
    "driver": "com.mysql.cj.jdbc.Driver"
}

# Table name in MySQL
mysql_table = "employees"

# Write DataFrame to MySQL
df.write \
    .format("jdbc").option("url", mysql_url).option("dbtable", mysql_table).option("user", mysql_properties["user"]).option("password", mysql_properties["password"]) \
    .option("driver", mysql_properties["driver"]) \
    .mode("append").save()
```
