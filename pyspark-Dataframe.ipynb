```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
spark
```

Create pyspark data frame from list of rows

```python
from datetime import datetime, date
import pandas as pd
from pyspark.sql import Row

df = spark.createDataFrame([Row(a=1, b=2, c="Yuga",d = date(2025,3,9)),
                                Row(a=1, b=2, c="Rushi",d = date(2025,3,9)),
                                    Row(a=1, b=2, c="Luna",d = date(2025,3,9)),
                                        Row(a=1, b=2, c="Namsvi",d = date(2025,3,9))])
df.show()
```
